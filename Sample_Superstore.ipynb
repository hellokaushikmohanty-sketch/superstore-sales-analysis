```python
# V20. Executive Summary Dashboard

# Step 1: Import matplotlib
import matplotlib.pyplot as plt

# ------------------------------------------
# Step 2: Calculate KPI Metrics
# ------------------------------------------

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()

# ------------------------------------------
# Step 3: Prepare Data for Charts
# ------------------------------------------

# Monthly Sales
monthly_sales = (
    df.groupby(df["Order Date"].dt.month)["Sales"]
      .sum()
)

# Sales by Category
sales_category = (
    df.groupby("Category")["Sales"]
      .sum()
)

# Sales by Region
sales_region = (
    df.groupby("Region")["Sales"]
      .sum()
)

# ------------------------------------------
# Step 4: Create Dashboard Layout
# ------------------------------------------

fig = plt.figure(figsize=(16,10))

# ------------------------------------------
# Step 5: KPI Cards
# ------------------------------------------

plt.subplot(2,2,1)

plt.axis("off")

plt.text(0,0.8,f"Total Sales\n${total_sales:,.2f}",
         fontsize=18)

plt.text(0,0.45,f"Total Profit\n${total_profit:,.2f}",
         fontsize=18)

plt.text(0,0.1,f"Total Orders\n{total_orders}",
         fontsize=18)

plt.title("Executive KPIs")

# ------------------------------------------
# Step 6: Monthly Sales Trend
# ------------------------------------------

plt.subplot(2,2,2)

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

# ------------------------------------------
# Step 7: Sales by Category
# ------------------------------------------

plt.subplot(2,2,3)

plt.bar(
    sales_category.index,
    sales_category.values
)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

# ------------------------------------------
# Step 8: Sales by Region
# ------------------------------------------

plt.subplot(2,2,4)

plt.bar(
    sales_region.index,
    sales_region.values
)

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

# ------------------------------------------
# Step 9: Adjust Layout
# ------------------------------------------

plt.tight_layout()

# ------------------------------------------
# Step 10: Display Dashboard
# ------------------------------------------

plt.show()
```
